To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
import os
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
# Constants
MODEL_LLAMA = 'llama3.2:latest'
OLLAMA_URL = "http://localhost:11434/v1"
```

```python
# Set up
ollama = OpenAI(base_url=OLLAMA_URL, api_key="ollama")
```

```python
question = """
Please explain what this code does and why:
"""
```

```python
# Prompts
system_prompt = "You are a technical assistant that provides specific and concise answer and explanation to a given technical question.\
      You return all your answers in markdown format."

user_prompt = "For the following question, please provide the technical details and explain the workings and objectives of the code:"
```

```python
# Allow user to write a specific code for question
user_question = input()
```

```python
# View input
user_question
```

```python
# Default question if no specific question added
if not user_question:
    user_question = """yield from {book.get("author") for book in books if book.get("author")}"""
```

```python
formatted_user_prompt = user_prompt + f"\n{user_question}"
print(formatted_user_prompt)
```

```python
query = ollama.chat.completions.create(
    model = MODEL_LLAMA,
    messages=[
        {"role":"system", "content":system_prompt},
        {"role":"user", "content": formatted_user_prompt}
    ],
    response_format={"type": "markdown"}
)

answer = query.choices[0].message.content
```

```python
display(Markdown(answer))
```
